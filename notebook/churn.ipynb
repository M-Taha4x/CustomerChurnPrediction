__ThresHold Tuning__

```python
y_pred_bal=balanced_model.predict(x_test_final)
```

```python
y_prob=balanced_model.predict_proba(x_test_final)[:,1]
```

```python
y_prob[:10]
```

```python
def evaluate_threshold(y_true,y_prob,threshold):
    y_pred=(y_prob>=threshold).astype(int)

    precision=precision_score(y_true,y_pred)
    recall=recall_score(y_true,y_pred)
    f1=f1_score(y_true,y_pred)
    return precision,recall,f1

thresholds=[0.3,0.4,0.5,0.6,0.7]
for threshold in thresholds:
    precision,recall,f1=evaluate_threshold(
        y_test,
        y_prob,
        threshold
    )
    print(
        f"Threshold: {threshold} "
        f"Precision: {precision:.4f} | "
        f"Recall: {recall:.4f} |"
        f"F1: {f1:.4f}"
    )
```

__SEARCHING FOR Best Thresholds__

```python
th=np.arange(0.1,0.91,0.01)

best_th=0
best_f1=0
for threshold in th:
    y_pred_th=(y_prob>=threshold).astype(int)
    f1=f1_score(y_test,y_pred_th)

    if f1>best_f1:
        best_f1=f1
        best_th=threshold

print(f'Best Threshold: {best_th:.4f}')
print(f"Best F1: {best_f1:.4f}")
```

```python
tH=0.65
y_pred_65=(y_prob>=tH).astype(int)
print("Precision: ",precision_score(y_test,y_pred_65))
print("Recall: ",recall_score(y_test,y_pred_65))
print("F1: ",f1_score(y_test,y_pred_65))
```

__Cross-Validation & Final Model Selection__

Building Preprocessing Properly

```python
categ_col=[
    "Senior Citizen",
    'Partner',
    'Dependents',
    'Internet Service',
    'Online Security',
    'Online Backup',
    'Device Protection',
    'Tech Support',
    'Contract',
    'Paperless Billing',
    'Payment Method'
]
```

```python
numeric_col=[
    'Tenure Months',
    'Monthly Charges',
    'Total Charges',
    'CLTV'
]
```

```python
from sklearn.compose import ColumnTransformer
```

```python
preprocessor=ColumnTransformer(
    transformers=[
        (
            'num',
            StandardScaler(),
            numeric_col
        ),
        (
            'col',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=False
            ),
            categ_col
        )
    ]
)
```

__Creating a Pipeline__

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
bal_lr_pipe=Pipeline(
    steps=[
        ('preprocessor',preprocessor),
        (
            'model',
            LogisticRegression(
                max_iter=1000,
                class_weight='balanced',
                random_state=42
            )
        )
    ]
)
```

Raw X
 ↓
ColumnTransformer
 ↓
Encoding + Scaling
 ↓
Logistic Regression
 ↓
Prediction

__CROSS VALIDATION__

```python
from sklearn.model_selection import StratifiedKFold,cross_validate
```

```python
#cv=Cross Validation
cv=StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
scoring={
    'accuracy':"accuracy",
    'precision':'precision',
    'recall':'recall',
    'f1':'f1',
    'roc_auc':'roc_auc'
}
```

```python
cv_result=cross_validate(
    bal_lr_pipe,
    x_train,
    y_train,
    cv=cv,
    scoring=scoring
)
```

```python
for metric in scoring:
    scores=cv_result[f'test_{metric}']
    print(f'{metric}:{scores.mean():.4f}±{scores.std():.4f}')
```
